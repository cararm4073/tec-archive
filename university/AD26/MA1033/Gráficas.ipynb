Cálculo de intervalo de tiempo $(t)$, usamos el periodo $(T)$

$$T = \frac{2\pi}{\omega} = \frac{2\pi}{\sqrt{40}} \approx 0.993 \text{ s}$$

```python
%matplotlib inline
```

```python
# @title Módulos
import numpy as np
import matplotlib.pyplot as plt
```

```python
# @title Parámetros y constantes
omega = np.sqrt(40)
```

```python
# @title Crear el arreglo para el tiempo
# tiempo (t) de 0 a 3 segundos
t = np.linspace(0, 3, 500)
```

```python
# @title Ecuaciones obtenidas
# posición, velocidad y aceleración
x = 0.10 * np.cos(omega * t)
v = -0.10 * omega * np.sin(omega * t)
a = -(omega**2) * 0.10 * np.cos(omega * t)
```

___
# Gráficas

```python
# @title Configuración
fig, axs = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Oscilador Armónico Simple', fontsize=18)

#Posición vs tiempo
axs[0, 0].plot(t, x, color='blue', linewidth=2)
axs[0, 0].set_title('1. Posición vs Tiempo')
axs[0, 0].set_xlabel('Tiempo (s)')
axs[0, 0].set_ylabel('Posición (m)')
axs[0, 0].grid(True)
axs[0, 0].axhline(0, color='red', linewidth=1)

#Velocidad vs tiempo
axs[0, 1].plot(t, v, color='green', linewidth=2)
axs[0, 1].set_title('2. Velocidad vs Tiempo')
axs[0, 1].set_xlabel('Tiempo (s)')
axs[0, 1].set_ylabel('Velocidad (m/s)')
axs[0, 1].grid(True)
axs[0, 1].axhline(0, color='red', linewidth=1)

#Aceleración vs tiempo
axs[1, 0].plot(t, a, color='orange', linewidth=2)
axs[1, 0].set_title('3. Aceleración vs Tiempo')
axs[1, 0].set_xlabel('Tiempo (s)')
axs[1, 0].set_ylabel('Aceleración (m/s²)')
axs[1, 0].grid(True)
axs[1, 0].axhline(0, color='red', linewidth=1)

#Velocidad vs posición
axs[1, 1].plot(x, v, color='purple', linewidth=2)
axs[1, 1].set_title('4. Velocidad vs Posición')
axs[1, 1].set_xlabel('Posición (m)')
axs[1, 1].set_ylabel('Velocidad (m/s)')
axs[1, 1].grid(True)
axs[1, 1].axhline(0, color='red', linewidth=1)
axs[1, 1].axvline(0, color='black', linewidth=1)

#MOstrar gráfica
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
